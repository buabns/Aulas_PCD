```python
# uso da biblioteca pandas
import pandas as pd
# nomes dos campos
nomes = ["fixed acidity", "volatile acidity", "citric acid", "residual sugar", "chlorides", "free sulfur dioxide", "total sulfur dioxide", "density", "pH","sulphates", "alcohol", "quality"]

df = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv',
skiprows = 1,
sep = ";",
names = nomes)

print(f'\n {df}')


```

```python
def classificar_nota(nota):
    if nota <= 4:
        return 'Baixa'
    elif nota in [5, 6]:
        return 'Media'
    else:
        return 'Alta'

# Aplica a função para criar a nova coluna de classificação
df['classificacao'] = df['quality'].apply(classificar_nota)

# Exibe a contagem de cada classificação e as primeiras linhas do DataFrame
print("Distribuição das classificações:")
print(df['classificacao'].value_counts())
print("\nDataFrame com a nova classificação:")
display(df.head(50))
```

```python
# Separando o dataset
# X será a matriz de características
X = df[df.columns.difference(['classificacao'])].values
print (f'X: {X}')

# y será o vetor com a classificação ou algo (target)
y = df['classificacao'].values
print (f'Y: {y}')
```

```python
# importamos a biblioteca sklearn que possui a implementação do algortimo
from sklearn.neighbors import KNeighborsClassifier
# definimos o numero de vizinhos = 3 para classificação
classificador = KNeighborsClassifier(n_neighbors = 7)
# uso do método fit para treinar o modelo - passando os valores do dataset
classificador.fit(X, y)
```

```python
# Desempenho do KNN
# Usando validação cruzada para
from sklearn.model_selection import cross_val_score
# quebra o dataset em 5 partes e usa uma delas para teste (validação)
scores_dt = cross_val_score(classificador, X, y, scoring='accuracy', cv=7)
# o testes é aletório, pegamos os valores médios
print(scores_dt.mean())
```

```python
# Treinando e avaliando - alterando o numero de vizinhos
for i in range(1,11):
    classificador = KNeighborsClassifier(n_neighbors = i)
    classificador.fit(X, y)
    scores_dt = cross_val_score(classificador, X, y, scoring='accuracy', cv=5)
    print(f'Vizinhos: {i}  Acurácia: {scores_dt.mean()}')
```

```python
# Cada vetor interno agora tem 12 valores numéricos correspondentes aos dados do vinho
novos_exemplos = [
    [7.4, 0.70, 0.00, 1.9, 0.076, 11.0, 34.0, 0.9978, 3.51, 0.56, 9.4, 5],
    [7.8, 0.88, 0.00, 2.6, 0.098, 25.0, 67.0, 0.9968, 3.20, 0.68, 9.8, 5]
]

# Avaliando a classificação
print(classificador.predict(novos_exemplos))
```
