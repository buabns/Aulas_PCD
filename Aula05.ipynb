```python
import json
import requests

#requisição via web para API
#resposta é um objeto - todas as possibilidades em https://requests.readthedocs.io/en/latest
resposta = requests.get("https://api.randomuser.me/")

#print se a requisição foi ok
print (resposta.status_code)

#print do texto da requisição
print (resposta.text)

#parse dos dados requisitados no formato json
dados = json.loads(resposta.text)

# print: dados brutos dos dicionário

# print: uma forma mais interessante de mostrar o JSON
print(json.dumps(dados, indent = 4, sort_keys=True))
```
