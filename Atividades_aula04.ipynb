```python
#EXERCICIO 1
numero = int(input("Digite um número: "))

if numero <= 1:
    print(f"O número {numero} não é primo.")
elif numero == 2:
    print(f"O número {numero} é primo.")
elif numero % 2 == 0:
    print(f"O número {numero} não é primo.")
else:
  # UTILIZEI IA NESSA PARTE AQUI DE BAIXO, MAS ENTENDI O FUNCIONAMENTO DO CODIGO
  # PROVAVELMENTE PODERIA TER SIDO ESCRITO DE JEITO MAIS SIMPLES PELA AI, MAS ENTENDI
    e_primo = True
    for i in range(3, numero, 2):
        if numero % i == 0:
            e_primo = False

    if e_primo:
        print(f"O número {numero} é primo.")
    else:
        print(f"O número {numero} não é primo.")
```

```python
#EXERCICIO 2
#TAMBEM UTILIZEI IA E ENTENDI O CONCEITO
import math

numero = float(input("Digite o número para calcular a raiz quadrada: "))

if numero < 0:
    print("Não é possível calcular a raiz quadrada de um número negativo.")
elif numero == 0:
    print(f"A raiz quadrada de {numero} é 0.0")
else:

    b = numero / 2.0
    if b == 0:
        b = 1.0

    precisao = 0.0001


    while abs(numero - (b * b)) > precisao:
        p = (b + numero / b) / 2
        b = p

    print(f"A raiz quadrada aproximada de {numero} (Método de Newton) é: {b:.10f}")
    print(f"A raiz quadrada real de {numero} (função math.sqrt) é: {math.sqrt(numero):.10f}")
```
