```python
import google_colab_selenium as gs
from selenium.webdriver.common.by import By
from time import sleep
import json

navegador = gs.Chrome()
```

### Acessar o site

Primeiro, vamos informar o endereço da página de pesquisa da Steam e abrir o site no navegador. Os parâmetros do endereço deixam a página em português e mostram os preços em reais. O sleep(4) dá alguns segundos para os resultados carregarem.

```python
url = "https://store.steampowered.com/search/?filter=topsellers&l=brazilian&cc=br"

navegador.get(url)
sleep(4)

print("Página:", navegador.title)
```

### Coletar as informações dos jogos

Vamos localizar os elementos que representam os jogos na página. Para cada jogo, vamos pegar o nome, a data de lançamento, o preço e o link. Essas informações serão guardadas em um dicionário e adicionadas a uma lista.

```python
jogos = []

itens = navegador.find_elements(By.CSS_SELECTOR, "a.search_result_row")

for item in itens:
    nome = item.find_element(By.CSS_SELECTOR, "span.title").text.strip()

    data = item.find_elements(By.CSS_SELECTOR, ".search_released")
    if data:
        data_lancamento = data[0].text.strip()
    else:
        data_lancamento = "Não informado"

    valor = item.find_elements(By.CSS_SELECTOR, ".discount_final_price")
    if valor:
        preco = valor[0].text.strip()
    else:
        preco = "Não informado"

    link = item.get_attribute("href")

    jogo = {
        "nome": nome,
        "data_lancamento": data_lancamento,
        "preco": preco,
        "link": link
    }

    jogos.append(jogo)

print("Jogos encontrados:", len(jogos))

for jogo in jogos:
    print(jogo)
```

### Salvar os dados em JSON

Agora vamos salvar a lista de jogos em um arquivo JSON. Esse formato organiza os dados em uma estrutura com os campos de cada jogo.

```python
with open("jogos_steam.json", "w", encoding="utf-8") as arquivo:
    json.dump(jogos, arquivo, ensure_ascii=False, indent=4)

print("O Arquivo foi salvo!")
```

### Baixar os arquivos

Por último, vamos baixar os arquivos gerados para o computador.

```python
from google.colab import files

files.download("jogos_steam.json")
```

### Sobre a coleta

A coleta foi feita na página de jogos mais vendidos da Steam, usando o Selenium para abrir o site e localizar os elementos dos jogos. Foram armazenados quatro campos: nome, data de lançamento, preço e link. Os mesmos dados foram salvos nos formatos JSON.

Os dados refletem o que estava aparecendo na página no momento da execução. Por isso, os preços e a lista de jogos podem mudar.
