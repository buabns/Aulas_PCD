```python
nome = str (input ("Fale o nome do aluno: "))
nota1 = float (input ("Fale a nota 1: "))
nota2 = float (input ("Fale a nota 2: "))
nota3 = float (input ("Fale a nota 3: "))
media = (nota1 + nota2 + nota3) / 3
print ("O aluno:", nome, "possui a media:")
print ("nota 1:", nota1, "nota 2:", nota2, "nota 3:1", nota3, "Sua media é: ", media)
```
