```python
#DESAFIO 1
alunos = int(input("Quantos alunos sua turma possui? "))

notas = []

for aluno in range(1, alunos + 1):
    nota = int(input(f"Qual a nota do {aluno} aluno? "))
    while nota > 10 or nota < 0:
        print("Forneça uma nota entre 0 e 10!")
        nota = int(input(f"Qual a nota do {aluno} aluno? "))
    notas.append(nota)

def maior_nota(notas):
    return max(notas)

def menor_nota(notas):
    return min(notas)

def media(notas):
    return sum(notas) / len(notas)

def aprovados(notas):
    quantidade = 0

    for nota in notas:
        if nota >= 6:
            quantidade += 1
    return quantidade

def reprovados(notas):
    quantidade = 0

    for nota in notas:
        if nota < 6:
            quantidade += 1
    return quantidade

print()
print("Resultado da turma")
print(f"Maior nota: {maior_nota(notas)}")
print(f"Menor nota: {menor_nota(notas)}")
print(f"Média da turma: {media(notas):.2f}")
print(f"Aprovados: {aprovados(notas)}")
print(f"Reprovados: {reprovados(notas)}")

```

```python
#DESAFIO 2
produtos = {}


def cadastrar_produto():
    codigo = int(input("Código do produto: "))
    nome = input("Nome do produto: ")
    preco = float(input("Preço do produto: R$ "))
    quantidade = int(input("Quantidade em estoque: "))
    produtos[codigo] = {
        "nome": nome,
        "preco": preco,
        "quantidade": quantidade
    }

    print("Produto cadastrado com sucesso")


def consultar_produto():
    codigo = int(input("Digite o código do produto: "))
    if codigo in produtos:
        produto = produtos[codigo]
        print(f"\nNome: {produto['nome']}")
        print(f"Preço: R$ {produto['preco']:.2f}")
        print(f"Quantidade: {produto['quantidade']}")
    else:
        print("Produto não encontrado.")


def valor_total_estoque():
    total = 0
    for produto in produtos.values():
        total += produto["preco"] * produto["quantidade"]
    print(f"Valor total do estoque: R$ {total:.2f}")


def maior_valor_unitario():
    if len(produtos) == 0:
        print("Nenhum produto cadastrado.")
        return

    produto_maior = max(
        produtos.values(),
        key=lambda produto: produto["preco"]
    )

    print(f"Produto de maior valor unitário: {produto_maior['nome']}")
    print(f"Preço: R$ {produto_maior['preco']:.2f}")


while True:

    print("    MENU   ")
    print("1 - Cadastrar produto")
    print("2 - Consultar produto")
    print("3 - Calcular valor total do estoque")
    print("4 - Mostrar produto de maior valor unitário")
    print("5 - Sair")

    opcao = int(input("Escolha uma opção: "))

    if opcao == 1:
        cadastrar_produto()

    elif opcao == 2:
        consultar_produto()

    elif opcao == 3:
        valor_total_estoque()

    elif opcao == 4:
        maior_valor_unitario()

    elif opcao == 5:
        print("Programa encerrado.")
        break

    else:
        print("Opção inválida.")
```
