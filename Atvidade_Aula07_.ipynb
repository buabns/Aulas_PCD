```python
import math
class Calculadora:

    def __init__(self, valor1: int, valor2: int):
        self.valor1 = valor1
        self.valor2 = valor2


    def setvalor1(self, valor1: int):
        self.valor1 = valor1

    def getvalor1(self):
        return self.valor1

    def setvalor2(self, valor2: int):
        self.valor2 = valor2

    def getvalor2(self):
        return self.valor2

    def soma(self):
        return self.valor1 + self.valor2

    def subtracao(self):
        return self.valor1 - self.valor2

    def multiplicacao(self):
        return self.valor1 * self.valor2

    def divisao(self):
        return self.valor1 / self.valor2

    def raiz(self):
        return print (f"a raiz é do valor 1!, que é: {math.sqrt(self.valor1)}")

calculadora = Calculadora(valor1 = 4, valor2 = 2)

calculadora.raiz()

```
