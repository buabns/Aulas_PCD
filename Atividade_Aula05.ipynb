```python
import json

def criar():

    status = [input(f"{campo}: ") for campo in ["Nome", "Idade", "Profissão", "Hobby"]]

    with open("personagem.json", "w") as f:
        json.dump(status, f)

def ler():
    with open("personagem.json", "r") as f:
        s = json.load(f)

    print(f"\n--- Perfil ---\nNome: {s[0]}\nIdade: {s[1]}\nProfissão: {s[2]}\nHobby: {s[3]}")


criar()
ler()
```
