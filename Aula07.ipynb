```python
class Calculadora:

    def __init__(self, valor1: int, valor2: int):
        self.valor1 = valor1
        self.valor2 = valor2


    def setvalor1(self, valor1: int):
        self.valor1 = valor1

    def getvalor1(self):
        return self.valor1

    def setvalor2(self, valor2: int):
        self.valor2 = valor2

    def getvalor2(self):
        return self.valor2

    def soma(self):
        return self.valor1 + self.valor2

    def subtracao(self):
        return self.valor1 - self.valor2

calculadora = Calculadora(valor1 = 11, valor2 = 11)

calculadora.soma()
calculadora.subtracao()
```
