```python

#------ Código Inicial

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

dados = {
    "Mes": ["Jan", "Fev", "Mar", "Abr", "Mai", "Jun"],
    "Notebook": [20, 25, 22, 30, 35, 40],
    "Mouse": [50, 55, 48, 60, 65, 70],
    "Teclado": [30, 28, 35, 40, 38, 45]
}

df = pd.DataFrame(dados)
print(df)

#----------

```

```python
# 1. Exploração
print(df)
print()
df.info()
print()
df.describe()
```

```python
# 2. Indicadores
medinote = np.mean(dados['Notebook'])
medimouse = np.mean(dados['Mouse'])
meditecla = np.mean(dados['Teclado'])
print(f"A media de vendas dos Notebooks é: {medinote:.2f}")
print(f"A media de vendas dos Mouses é: {medimouse:.2f}")
print(f"A media de vendas dos Teclados é: {meditecla:.2f}")
print()

maxnote = np.max(dados['Notebook'])
maxmouse = np.max(dados['Mouse'])
maxtecla = np.max(dados['Teclado'])
print(f"O maior valor de vendas dos Notebooks é: {maxnote:.2f}")
print(f"O maior valor de vendas dos Mouses é: {maxmouse:.2f}")
print(f"O maior valor de vendas dos Teclados é: {maxtecla:.2f}")
print()

minnote = np.min(dados['Notebook'])
minmouse = np.min(dados['Mouse'])
mintecla = np.min(dados['Teclado'])
print(f"O menor valor de vendas dos Notebooks é: {minnote:.2f}")
print(f"O menor valor de vendas dos Mouses é: {minmouse:.2f}")
print(f"O menor valor de vendas dos Teclados é: {mintecla:.2f}")

```

```python
# 3. Evolução temporal
plt.figure(figsize=(8, 5))

plt.plot(df["Mes"], df["Notebook"], marker="o", label="Notebook")
plt.plot(df["Mes"], df["Mouse"], marker="o", label="Mouse")
plt.plot(df["Mes"], df["Teclado"], marker="o", label="Teclado")

plt.title("Evolução das Vendas")
plt.xlabel("Mês")
plt.ylabel("Quantidade Vendida")

plt.legend()
plt.show()
```

```python
# 4. Comparação
totais = {
    "Notebook": df["Notebook"].sum(),
    "Mouse": df["Mouse"].sum(),
    "Teclado": df["Teclado"].sum()
}

plt.figure(figsize=(7, 5))
plt.bar(totais.keys(), totais.values(), color=["blue", "orange", "green"])


plt.title("Total de Vendas por Produto")
plt.xlabel("Produto")
plt.ylabel("Total de Unidades Vendidas")

plt.show()
```

```python
# 5. Distribuição

plt.figure(figsize=(7, 5))
plt.hist(df["Notebook"], bins=5, color="orange", edgecolor="black", alpha=0.7)

plt.title("Distribuição das Vendas Mensais de Notebooks")
plt.xlabel("Quantidade Vendida")
plt.ylabel("Frequência (Meses)")

plt.show()
```

```python
# 6. Interpretação

maior_produto = max(totais, key=totais.get)
print(f"Produto mais vendido: {maior_produto} ({totais[maior_produto]} unidades)")

for produto in ['Notebook', 'Mouse', 'Teclado']:
    idx_max = df[produto].idxmax()
    mes_max = df.loc[idx_max, 'Mes']
    val_max = df.loc[idx_max, produto]
    print(f"Maior venda de {produto}: {mes_max} ({val_max} unidades)")

print("\nConclusão baseada nos gráficos:")
print("1. O Mouse é o produto com melhor desempenho constante, dominando as vendas.")
print("2. Há uma clara tendência de alta para todos os produtos ao longo dos meses, visível no gráfico de linhas.")
print("3. O mês de Junho se consolidou como o melhor mês de vendas para todos os três itens analisados.")
```

```python
# 7. Venda de teclados

# Motivo desse grafico: Permite comparar as vendas mensais lado a lado e ver na hora quais meses ficaram abaixo ou acima da média geral.

media_teclado = df["Teclado"].mean()
plt.figure(figsize=(7, 4))
plt.bar(df["Mes"], df["Teclado"], color="skyblue", edgecolor="black")
plt.axhline(media_teclado, color="red", linestyle="--", linewidth=2, label=f"Média ({media_teclado:.0f} un)")

plt.title("Vendas Mensais de Teclados")
plt.xlabel("Mês")
plt.ylabel("Unidades Vendidas")
plt.legend()
plt.show()

# Informações observadas:
# 1. Os primeiros três meses (Jan, Fev, Mar) ficaram abaixo da média (36 unidades).
# 2. A partir de Abril, as vendas superaram a média, mostrando uma tendência de crescimento.
```
