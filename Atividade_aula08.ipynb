```python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/andr3nf/clima_municipios_brasileiros/refs/heads/main/clima_historico_municipal.csv')

# 1. Localizar e mostrar todas as informações de Ubiretama
cidade = df[df['nome_municipio'] == 'Ubiretama'].iloc[0]
print("--- TODAS AS INFORMAÇÕES DE UBIRETAMA ---")
display(cidade.to_frame())

# 2. Mês de temperatura mínima
temp_cols = [c for c in df.columns if 'temp_media_' in c]
mes_min_temp = cidade[temp_cols].idxmin().replace('temp_media_', '')
print(f"\nMês de temperatura mínima: {mes_min_temp.upper()} ({cidade[temp_cols].min():.2f}°C)")

# 3. Mês de maior déficit hídrico
def_cols = [c for c in df.columns if 'def_' in c]
mes_max_def = cidade[def_cols].idxmax().replace('def_', '')
print(f"Mês de maior déficit hídrico: {mes_max_def.upper()} ({cidade[def_cols].max():.2f} mm)")

# 4. Média de temperatura pelas 4 estações do ano
estacoes = {
    'Verão': ['dez', 'jan', 'fev'],
    'Outono': ['mar', 'abr', 'mai'],
    'Inverno': ['jun', 'jul', 'ago'],
    'Primavera': ['set', 'out', 'nov']
}

print("\nMédia de temperatura por estação:")
for estacao, meses in estacoes.items():
    cols = [f'temp_media_{m}' for m in meses]
    media = cidade[cols].mean()
    print(f"  {estacao}: {media:.2f}°C")
```
