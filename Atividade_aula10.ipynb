```python
import numpy as np
dados = np.array([
                   [120, 15000, 100, 125.00, 4],
                   [135, 16200, 108, 120.00, 5],
                   [98, 11700, 82, 119.39, 3],
                   [150, 21000, 125, 140.00, 7],
                   [175, 24500, 140, 140.00, 9],
                   [160, 22400, 132, 140.00, 6],
                   [110, 13200, 95, 120.00, 2]
])
                                    #OBS: EXECUTE TODAS AS CÉLULAS PARA FUNCIONAR, CÉLULA POR CéLULA NÃO IRA FUNCIONAR!!
 #PARTE 1
dias = np.array([
"Segunda",
"Terça",
"Quarta",
"Quinta",
"Sexta",
"Sábado",
"Domingo"
])

print(f"Quantidade de linhas e colunas: {dados.shape}")
print(f"Valor total de valores: {dados.size}")
print(f"Número de dimensões do array: {dados.ndim}")
print(f"Tipos de dados armazenados: {dados.dtype}")
```

```python
#PARTE 2
print("Sexta-feira:", dados[4])
print("Faturamento:", dados[:, 1])
print("Reclamações:", dados[:, 4])
print("Quinta a Domingo:\n", dados[3:])
print("Pedidos, Faturamento e Clientes:\n", dados[:, :3])
```

```python
#PARTE 3
total_pedidos = np.sum(dados[:, 0])
faturamento_total = np.sum(dados[:, 1])
total_clientes = np.sum(dados[:, 2])
media_pedidos = np.mean(dados[:, 0])
media_faturamento = np.mean(dados[:, 1])
media_reclamações = np.mean(dados[:, 4])

print(f"Total de pedidos: {total_pedidos}")
print(f"Faturamento total: R$ {faturamento_total:.2f}")
print(f"Total de clientes: {total_clientes}")
print(f"Média diária de pedidos: {media_pedidos:.2f}")
print(f"Média diária de faturamento: R$ {media_faturamento:.2f}")
print(f"Média de reclamações por dia: {media_reclamações:.2f}")
```

```python
#PARTE 4
dia_max_pedidos = dias[np.argmax(dados[:, 0])]
dia_min_pedidos = dias[np.argmin(dados[:, 0])]
print(f"Maior número de pedidos: {dia_max_pedidos}")
print(f"Menor número de pedidos: {dia_min_pedidos}")

dia_max_fat = dias[np.argmax(dados[:, 1])]
dia_min_fat = dias[np.argmin(dados[:, 1])]
print(f"Maior faturamento: {dia_max_fat}")
print(f"Menor faturamento: {dia_min_fat}")

amplitude_fat = np.max(dados[:, 1]) - np.min(dados[:, 1])
print(f"Amplitude do faturamento: R$ {amplitude_fat:.2f}")
```

```python
#PARTE 5
mask_fat = dados[:, 1] > 20000
print("Dias com faturamento > R$ 20.000:", dias[mask_fat])

mask_ped = dados[:, 0] > 150
print("Dias com mais de 150 pedidos:", dias[mask_ped])

mask_rec = dados[:, 4] > 5
print("Dias com mais de 5 reclamações:", dias[mask_rec])

print("Faturamentos (> 150 pedidos):", dados[dados[:, 0] > 150, 1])

dias_abaixo_media = np.sum(dados[:, 1] < np.mean(dados[:, 1]))
print(f"Dias abaixo da média de faturamento: {dias_abaixo_media}")
```

```python
#PARTE 6
classificacao = np.where(
    dados[:, 1] >= 20000,
    "Alto desempenho",
    "Desempenho normal"
)
print("Classificação dos dias:", classificacao)

qtd_alto_desempenho = np.sum(classificacao == "Alto desempenho")
print(f"Dias de alto desempenho: {qtd_alto_desempenho}")
```

```python
#PARTE 7
fat = dados[:, 1]
print(f"Média: R$ {np.mean(fat):.2f}")
print(f"Mediana: R$ {np.median(fat):.2f}")
print(f"Desvio padrão: R$ {np.std(fat):.2f}")
print(f"Mínimo: R$ {np.min(fat):.2f}")
print(f"Máximo: R$ {np.max(fat):.2f}")
```

```python
#PARTE 8
print("Média de cada indicador:", np.mean(dados, axis=0))
print("Maior valor de cada indicador:", np.max(dados, axis=0))
print("Menor valor de cada indicador:", np.min(dados, axis=0))
```

```python
#PARTE 9
melhordia = dias[np.argmax(dados[:, 1])]
maior_ped_dia = np.argmax(dados[:, 0])
maior_fat_dia = np.argmax(dados[:, 1])
maior_rec_dia = np.argmax(dados[:, 4])
mask_integrada = (dados[:, 0] > 150) & (dados[:, 1] > 20000)

print(f"Melhor dia: {melhordia}")
print(f"O dia de maior faturamento é o mesmo de maior pedidos? {maior_ped_dia == maior_fat_dia}")
print(f"O dia de maior faturamento é o mesmo de maior reclamações? {maior_fat_dia == maior_rec_dia}")
print(f"Dias simultaneamente com >150 pedidos e faturamento >20k: {np.sum(mask_integrada)}")
```

```python
#PARTE 10
total_pedidos = np.sum(dados[:, 0])
faturamento_total = np.sum(dados[:, 1])
media_pedidos = np.mean(dados[:, 0])
media_faturamento = np.mean(dados[:, 1])
melhor_dia = dias[np.argmax(dados[:, 1])]
pior_dia = dias[np.argmin(dados[:, 1])]
max_reclamacoes = np.max(dados[:, 4])
dias_alto_desempenho = np.sum(dados[:, 1] >= 20000)

print("=== RESUMO DA SEMANA ===")
print(f"Total de pedidos: {total_pedidos}")
print(f"Faturamento total: R$ {faturamento_total:.2f}")
print(f"Média diária de pedidos: {media_pedidos:.2f}")
print(f"Média diária de faturamento: R$ {media_faturamento:.2f}")
print(f"Melhor dia: {melhor_dia}")
print(f"Pior dia: {pior_dia}")
print(f"Maior número de reclamações: {max_reclamacoes}")
print(f"Dias de alto desempenho: {dias_alto_desempenho}")
```
